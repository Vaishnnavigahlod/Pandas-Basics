```python
import pandas as pd
```

```python

data = {
    "Student": ["Amit", "Neha", "Rahul", "Pooja"],
    "Marks": [85, 92, 76, 88],
    "City": ["Nagpur", "Pune", "Mumbai", "Nashik"]
}

df = pd.DataFrame(data, index=[101, 105, 110, 115])


```

```python
print(df)
```

```python
#first 2 row
df.iloc[0:2] #iloc me indexing 0 se start hoti hai
```

```python
#position 1 se 3
df.iloc[1:4]
```

```python
#row and column slicing
df.iloc[0:2 , 0:2]
```

```python
#last 2 row
df.iloc[2:4]
```

```python
#first 3 rows
df.iloc[:3]
```

```python
#1 se leke sari rows
df.iloc[1:]
```

```python
#ek specific column ki first 3 values
df.iloc[:3,1] # : 3---first 3 rows, 1----column ki possition
```
