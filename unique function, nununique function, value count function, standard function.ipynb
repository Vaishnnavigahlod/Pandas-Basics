```python
import pandas as pd
```

```python
data = {
    "Patient_ID": ["P101", "P102", "P103", "P104", "P105",
                   "P106", "P107", "P108", "P109", "P110"],

    "Age": [45, 62, 34, 55, 41, 68, 29, 52, 47, 73],

    "City": ["Hyderabad", "Warangal", "Hyderabad", "Vijayawada",
             "Hyderabad", "Nizamabad", "Hyderabad", "Warangal",
             "Hyderabad", "Karimnagar"],

    "Surgery_Type": [
        "Cholecystectomy", "CABG", "Appendectomy",
        "Hernia Repair", "Knee Replacement", "PTCA",
        "Appendectomy", "Cholecystectomy",
        "Hernia Repair", "Knee Replacement"
    ],
    "Operation_Duration_Min": [90, 240, 75, 110, 180,
                               150, 70, 95, 120, 200],

    "Length_of_Stay_Days": [2, 7, 1, 3, 5,
                            4, 1, 2, 3, 6],

    "Admission_Type": [
        "Planned", "Emergency", "Emergency", "Planned",
        "Planned", "Emergency", "Emergency", "Planned",
        "Planned", "Planned"
    ]
}

df = pd.DataFrame(data)
```

UNIQUE FUNCTION:

it returns all unique/different values from a column.syntax:df["column_name"].unique() ......

it gives value. repitation nhi honga koi value ka . single alag alag value denge. humare patients kin kin cities seaate hai. (kya cahiye)

```python
#unique function
df["City"].unique()
```

NUNUNIQUE:it returns the number of unique/difference values.syntax:df["column_name"].nunique()
 
it gives counting of value. it counts how many values are pressent.(kitne cahiye)

```python
#nunique()
df["City"].nunique()
```

VALUE COUNT: it counts how many times each value occures.syntax:df["column_name"].value_counts()

 har value kitne baar aa rhi hai uss ka count deta hai.(konse city se kitne patient aaye , kon kitne baar aaya vo batata hai ye) (how many times)

```python
df["City"].value_counts()
```

```python
df["Surgery_Type"].value_counts()
```
