```python
import pandas as pd
```

```python
data = {
    "Patient_ID": ["P101", "P102", "P103", "P104", "P105"],
    "Age": [45, 62, None, 55, 41],
    "City": ["Hyderabad", "Warangal", "Hyderabad", None, "Hyderabad"],
    "Operation_Duration_Min": [90, 240, 75, None, 180]
}

df = pd.DataFrame(data)
```

```python
print(df)
```

#missing value : none--- data available nhi hai

ISNULL :  it is used to check whether values are missing in a dataframe. syntax:df.isnull()
          output is (TRUE) --->value is missing
          output is (FALSE) --->value is available
          ye batata hai ki missing value kaha hai.(where is missing)

```python
#ex:isnull
print(df.isnull())
```

```python
#ex:isnull
print(df["Age"].isnull())
```

ISNULL().SUM() : it is used to count missing values. syntax: isnull().sum()
                 ye batata hai ki kitne missing values hai.(how many are missing).

```python
#kitne missing vales hai
print(df.isnull().sum()) #used to count missing values in each column
```
