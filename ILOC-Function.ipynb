```python
import pandas as pd

data = {
    "Patient_ID": ["P101", "P102", "P103", "P104", "P105",
                   "P106", "P107", "P108", "P109", "P110"],

    "Age": [45, 62, 34, 55, 41, 68, 29, 52, 47, 73],

    "City": ["Hyderabad", "Warangal", "Hyderabad", "Vijayawada",
             "Hyderabad", "Nizamabad", "Hyderabad", "Warangal",
             "Hyderabad", "Karimnagar"],

    "Surgery_Type": [
        "Cholecystectomy", "CABG", "Appendectomy",
        "Hernia Repair", "Knee Replacement", "PTCA",
        "Appendectomy", "Cholecystectomy",
        "Hernia Repair", "Knee Replacement"
    ],
    "Operation_Duration_Min": [90, 240, 75, 110, 180,
                               150, 70, 95, 120, 200],

    "Length_of_Stay_Days": [2, 7, 1, 3, 5,
                            4, 1, 2, 3, 6],

    "Admission_Type": [
        "Planned", "Emergency", "Emergency", "Planned",
        "Planned", "Emergency", "Emergency", "Planned",
        "Planned", "Planned"
    ]
}

df = pd.DataFrame(data)
```

```python
print(df)
```

Difference Between loc and iloc

```python
#index 2 wali patient ki complete information display
df.loc[2]
```

```python
#third patient ki complete information display cahiye (index starts with 0)
df.iloc[2]
```

```python
#what is the main difference between loc and iloc
#lable ----loc

#position index ---iloc
```

```python
#1)hospital wants to check the details of the 4th patient
df.iloc[3]
```

```python
#2)patient whoes index label is 7(loc)
df.loc[6]
```

```python
#3)display the first 5 patients, show only --patient_id, city, age
df.loc[0:5][["Patient_ID","City","Age"]]
```

```python
#4)last 3 patient data
df.iloc[-3:]
```

```python
#5)patient_id, city (loc)
df.loc[:,["Patient_ID","City"]] #sari row cahiye thi iss liye(:)use kiya, or 2 column cahiye the iss liye us ke 
#aage patient id or city likha
```

```python
#6)find all patients from warangal dispaly patient_id, surgery, age
df.loc[df["City"]=="Warangal",["Patient_ID","Surgery_Type","Age","City"]]
```

```python
#7)find patients whoes age is greater than 50. dispaly patient_id, age, city
df.loc[df["Age"]>50,["Patient_ID","Age","City"]]
```

```python
#8)rows 2 to 5, column 1 to 4
df.iloc[2:6, 1:5]
```

```python
#9)Patient at index 4, patient_id, surgery_type (indexing starts from 0 in iloc)
df.iloc[4][["Patient_ID","Surgery_Type"]]
```

```python
#challange: find the youngest emergency patient, display patient_id, age ,city
df.loc[df["Admission_Type"]=="Emergency","Age"].min()[["Patient_ID","Age","City"]]
```
