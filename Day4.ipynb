```python
import pandas as pd

data = {
    "Patient_ID": ["P101", "P102", "P103", "P104", "P105",
                   "P106", "P107", "P108", "P109", "P110"],

    "Age": [45, 62, 34, 55, 41, 68, 29, 52, 47, 73],

    "City": ["Hyderabad", "Warangal", "Hyderabad", "Vijayawada",
             "Hyderabad", "Nizamabad", "Hyderabad", "Warangal",
             "Hyderabad", "Karimnagar"],

    "Surgery_Type": [
        "Cholecystectomy", "CABG", "Appendectomy",
        "Hernia Repair", "Knee Replacement", "PTCA",
        "Appendectomy", "Cholecystectomy",
        "Hernia Repair", "Knee Replacement"
    ],
    "Operation_Duration_Min": [90, 240, 75, 110, 180,
                               150, 70, 95, 120, 200],

    "Length_of_Stay_Days": [2, 7, 1, 3, 5,
                            4, 1, 2, 3, 6],

    "Admission_Type": [
        "Planned", "Emergency", "Emergency", "Planned",
        "Planned", "Emergency", "Emergency", "Planned",
        "Planned", "Planned"
    ]
}

df = pd.DataFrame(data)

```

```python
print(df)
```

```python
#1)first 5 records
print(df.head())
```

```python
#2)number of rows and columns
print(df.shape)
```

```python
#3)column names
print(df.columns)
```

```python
#4)patient ID,city,surgery type, addmission type
print(df[["Patient_ID","City","Surgery_Type","Admission_Type"]])
```

```python
#5)Emergncy cases
print(df[df["Admission_Type"]=="Emergency"])
```

```python
#6)operations greater than 150 minutes
print(df[df["Operation_Duration_Min"]>150])
```

```python
#7)Hyderabad patients
print(df[df["City"]=="Hyderabad"][["Patient_ID","Age","Surgery_Type","Admission_Type"]])
```

```python
#Average operation duration
print(df["Operation_Duration_Min"].mean())
```

```python
#highest operation duration
print(df["Operation_Duration_Min"].max())
```

```python
#8)Patient with highest operation duration

print(df.loc[df["Operation_Duration_Min"].idxmax(),["Patient_ID","Surgery_Type","Operation_Duration_Min"]])
```

```python
#9)operation >100  minutes AND hospital stay >3 days
print(df[(df["Operation_Duration_Min"]>100)&(df["Length_of_Stay_Days"]>3)][["Patient_ID","Surgery_Type","Operation_Duration_Min","Length_of_Stay_Days"]])
```

```python
#10)emergency cases with operation >100 minutes
print(df[(df["Admission_Type"]=="Emergency")&(df["Operation_Duration_Min"]>100)][["Patient_ID","Surgery_Type","Operation_Duration_Min","Admission_Type"]])
```
